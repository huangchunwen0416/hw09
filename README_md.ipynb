```python
# HW09｜Python 資料型別上機實作：兩場活動報名整理器

- **姓名**：黃郡玟
- **學號**：D1483952

---

## 一、程式用途與執行方式

### 程式用途
本程式針對「桌遊體驗」（每人 120 元）與「電影欣賞」（每人 80 元）兩場活動的原始報名文字進行自動化整理。先透過字串方法去除多餘空白並統一轉換為小寫，再結合 Python 四種基礎容器處理資料：
- **`tuple`**：存放不可變動的活動名稱（`("桌遊體驗", "電影欣賞")`）。
- **`list`**：儲存清理後尚未去重的原始順序名單（檢視重複報名情況），以及排序切片（`[:2]`）後產生的前兩位通知名單。
- **`set`**：排除單場活動內的重複報名者，並利用交集（`&`）、聯集（`|`）與雙向差集（`-`）分析兩場活動的參加者重疊度。
- **`dict`**：記錄各場活動的單人票價，並統整最終人數與應收金額報表。

### 如何在 Colab 開啟並執行
1. 點選本儲存庫中的 `activity_signup_D1483952.ipynb` 檔案。
2. 點擊頁面上方的 **「Open in Colab」**（或下載 `.ipynb` 檔案後匯入 Google Colab）。
3. 進入 Colab 介面後，點選上方工具列的 **「執行階段」→「全部執行」**，即可依序檢視「測試一」、「測試二」與「測試三」的執行結果。

---

## 二、三組實測摘要

| 測試組別 | 桌遊人數 | 電影人數 | 共同人數 | 總不重複人數 | 報名人次 | 總應收 |
| :--- | :---: | :---: | :---: | :---: | :---: | :---: |
| **測試一：原始資料** | 3 人 | 3 人 | 1 人 | 5 人 | 6 人次 | 600.00 元 |
| **測試二：新增一位電影參加者** | 3 人 | 4 人 | 1 人 | 6 人 | 7 人次 | 680.00 元 |
| **測試三：兩場去重後完全相同** | 3 人 | 3 人 | 3 人 | 3 人 | 6 人次 | 600.00 元 |

---

## 三、問題回答

### 問題一：測試一中，整理後的桌遊串列有 4 筆資料，為什麼桌遊費用應以 3 人計算？直接把字串拆成 list，和再轉成 set，各自保留了什麼、失去了什麼？
- **費用以 3 人計算的原因**：經字串清理後的桌遊串列為 `['amy', 'ben', 'amy', 'cora']`，其中 `'amy'` 因為原始輸入大小寫與空格差異（`" Amy"` 與 `"amy"`）而被重複登記兩次。依照活動計費原則，同一人在同一場次重複報名僅視為 1 位有效參加者、收取一次費用，因此實際參與人數為 3 人（Amy、Ben、Cora），桌遊應收金額為 $3 \times 120 = 360$ 元。
- **直接拆成 `list`**：
  - **保留了**：參加者填寫報名的先後次序，以及每個人重複填寫的完整紀錄與次數。
  - **失去了**：無法直接得知實際去重後的有效人數，也無法直接進行交集、差集等集合比對。
- **再轉成 `set`**：
  - **保留了**：該場次所有不重複的獨立參加者名單，並能直接執行交集（`&`）、聯集（`|`）、差集（`-`）等名單分析。
  - **失去了**：原本報名的先後順序，以及個別參加者曾重複報名幾次的次數細節。

### 問題二：測試三中，兩場活動只有 3 位不同的人，為什麼是 6 人次、總應收仍是 600 元？請用集合關係與算式說明。
- **集合關係說明**：在測試三中，桌遊參加者集合 $A = \{\text{'amy'}, \text{'ben'}, \text{'cora'}\}$，電影參加者集合 $B = \{\text{'amy'}, \text{'ben'}, \text{'cora'}\}$，兩集合元素完全一致（$A = B$）。
  - **總不重複人數**計算的是兩場的**聯集**（$A \cup B$）：由於兩場名單完全重疊，聯集後仍為這 3 位同學，故總不重複人數 $|A \cup B| = 3$ 人。
  - **報名人次**計算的是兩場各自去重後的參與人數總和（$|A| + |B|$）：因每位同學皆同時報名兩場活動，每人各貢獻 2 人次。
- **算式說明**：
  - **報名人次**：$|A| + |B| = 3 + 3 = 6$ 人次。
  - **總應收金額**：兩場活動獨立收費，桌遊應收 $3 \times 120 = 360$ 元，電影應收 $3 \times 80 = 240$ 元，合計 $360 + 240 = 600$ 元。

---

## 四、AI 使用紀錄

- **使用工具**：Google Gemini
- **主要提示詞**：請協助建立 HW09「兩場活動報名整理器」的 Colab 筆記本與 GitHub README 說明文件，需運用 `tuple`、`list`、`set`、`dict` 與字串方法完成三組測試案例，且不使用迴圈與自訂函式。
- **個人修改與驗證內容**：
  1. 確認字串清理流程有先執行 `.strip().replace(" ", "").lower()` 再轉為 `set`，使大小寫與空白不同的重複名字（如 `" Amy"` 與 `"amy"`）能正確合併。
  2. 檢查費用計算字典 `report_data` 是採用去重集合的長度 `len(unique_bg)` 與 `len(unique_mv)` 乘以單價，避免誤用未去重串列長度導致重複收費。
  3. 實際在 Google Colab 執行全部三個測試區塊並核對輸出：確認測試二新增 Frank 後電影人數變為 4 人、總應收為 680.00 元；測試三兩場名單相同時，只參加桌遊與只參加電影的差集皆正確顯示為空串列 `[]`。
```
